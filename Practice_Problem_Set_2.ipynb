```python
# 2D Array-DS
import math
import os
import random
import re
import sys

#
# Complete the 'hourglassSum' function below.
#
# The function is expected to return an INTEGER.
# The function accepts 2D_INTEGER_ARRAY arr as parameter.
#

def hourglassSum(arr):
    # Initialize max_sum to the absolute minimum possible hourglass sum (-9 * 7 = -63)
    max_sum = -63

    # Check all possible 3x3 hourglass configurations in the 6x6 matrix
    for i in range(4):
        for j in range(4):
            # Calculate the sum for the current hourglass pattern
            current_sum = (
                arr[i][j] + arr[i][j+1] + arr[i][j+2] +    # Top row
                arr[i+1][j+1] +                            # Middle element
                arr[i+2][j] + arr[i+2][j+1] + arr[i+2][j+2]   # Bottom row
            )

            # Update max_sum if the current hourglass sum is larger
            if current_sum > max_sum:
                max_sum = current_sum

    return max_sum

if __name__ == '__main__':
    # In a typical Colab environment, OUTPUT_PATH is not defined.
    # We'll modify this section to read input and print output directly.

    arr = []

    print("Please enter the 6x6 array row by row. Each row should have 6 space-separated integers.")
    for _ in range(6):
        arr.append(list(map(int, input().rstrip().split())))

    result = hourglassSum(arr)

    print(result)
```

```python
# Sparse Arrays
#!/bin/python3

import math
import os
import random
import re
import sys
from collections import Counter

#
# Complete the 'matchingStrings' function below.
#
# The function is expected to return an INTEGER_ARRAY.
# The function accepts following parameters:
#  1. STRING_ARRAY stringList
#  2. STRING_ARRAY queries
#

def matchingStrings(stringList, queries):
    # Step 1: Count frequencies of each string inside stringList using a hash map
    counts = Counter(stringList)

    # Step 2: Map each query to its count from the map (defaulting to 0 if not found)
    return [counts[query] for query in queries]

if __name__ == '__main__':
    # In a typical Colab environment, OUTPUT_PATH is not defined.
    # We'll modify this section to read input and print output directly.

    print("Enter the number of strings in stringList:")
    stringList_count = int(input().strip())

    stringList = []
    print("Enter each string for stringList on a new line:")
    for _ in range(stringList_count):
        stringList_item = input()
        stringList.append(stringList_item)

    print("Enter the number of queries:")
    queries_count = int(input().strip())

    queries = []
    print("Enter each query string on a new line:")
    for _ in range(queries_count):
        queries_item = input()
        queries.append(queries_item)

    res = matchingStrings(stringList, queries)

    # Print the results directly to standard output
    for r in res:
        print(r)
```

```python
# Lapindromes
# cook your dish here
import sys

def main():
    input_data = sys.stdin.read().split()
    if not input_data:
        return

    t = int(input_data[0])
    results = []

    for i in range(1, t + 1):
        s = input_data[i]
        length = len(s)
        mid = length // 2

        left_half = s[:mid]
        right_half = s[mid:] if length % 2 == 0 else s[mid+1:]

        if sorted(left_half) == sorted(right_half):
            results.append("YES")
        else:
            results.append("NO")

    print('\n'.join(results))

if __name__ == '__main__':
    main()
```

```python
# best time to buy and sell stock
from typing import List

class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        min_price = float('inf')
        max_profit = 0

        for price in prices:
            if price < min_price:
                min_price = price
            elif price - min_price > max_profit:
                max_profit = price - min_price

        return max_profit
```

```python
# longest common prefix
class Solution:
    def longestCommonPrefix(self, strs: List[str]) -> str:
        if not strs:
            return ""

        prefix = strs[0]

        for i in range(1, len(strs)):
            while not strs[i].startswith(prefix):
                prefix = prefix[:-1]
                if not prefix:
                    return ""

        return prefix
```

```python
# search insert position
class Solution:
    def searchInsert(self, nums: List[int], target: int) -> int:
        low, high = 0, len(nums) - 1

        while low <= high:
            mid = (low + high) // 2
            if nums[mid] == target:
                return mid
            elif nums[mid] < target:
                low = mid + 1
            else:
                high = mid - 1

        return low
```
